```python
#language check

import requests

repo = "akavel/hellomello"
r = requests.get(f"https://api.github.com/repos/{repo}")
print(r.json().get('language'))
```

```python
# Languge check from json

import requests

repo = "akavel/hellomello"
url = f"https://api.github.com/repos/{repo}/languages"
r = requests.get(url)
print(r.json())
```

```python
import requests

url = "https://api.github.com/repos/Krosxx/yyets_flutter"

response = requests.get(url)
data = response.json()

print("Language:", data.get('language'))
print("Name:", data.get('name'))
print("Description:", data.get('description'))
print("Topics:", data.get('topics', []))  
```

```python
import requests

r = requests.get(
    "https://api.github.com/repos/Krosxx/yyets_flutter",
    headers={"Accept": "application/vnd.github.mercy-preview+json"}
)
print(r.json()["topics"])
```
