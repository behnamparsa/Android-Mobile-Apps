```python
import requests

# Correct API endpoint (not the HTML page)
#https://github.com/angryziber/picasa-gallery
#https://github.com/JetBrains/kotlin # tool
#https://github.com/JetBrains/intellij-plugin-verifier.git # tool
#https://github.com/fnberta/PopularMovies real mobile app
#https://github.com/Yumenokanata/FragmentManager # library
#https://github.com/008chen/InterpolatorShow
#https://github.com/nightmare-space/code_lfa.git
#https://github.com/0xchat-app/0xchat-app-main  android in readme only - flutter cross platform app
repo = "camcamfresh/Xiaomi-M365-BLE-Controller-Replacement"
url = f"https://api.github.com/repos/{repo}"



response = requests.get(url)
data = response.json()

print("Language:", data.get('language'))
print("Name:", data.get('name'))
print("Description:", data.get('description'))

# Topics are fetched with a custom Accept header
topics_url = f"https://api.github.com/repos/{repo}/topics"
headers = {"Accept": "application/vnd.github.mercy-preview+json"}
topics_response = requests.get(topics_url, headers=headers)

if topics_response.ok:
    print("Topics:", topics_response.json().get('names', []))
else:
    print("Topics: Not available")
```
