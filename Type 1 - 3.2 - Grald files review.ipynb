```python
import os
import pandas as pd

# === CONFIGURATION ===
GRADLE_DIR = r"C:\Android Mobile App\Step2_Clone_Repo\Type_1\July_31\Build_Files"
OUTPUT_CSV = r"C:\Android Mobile App\Step3_Instr_Testing_Analysis\Type_1\July_31\3.2_Gradle_List_ShallowC.csv"

# === TEST DETECTION LOGIC ===
def detect_tests(file_path):
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read().lower()
        has_unit_test = any(keyword in content for keyword in ['junit', 'testimplementation'])
        has_instrumentation_test = any(keyword in content for keyword in ['androidtestimplementation', 'espresso', 'uiautomator'])
        return has_unit_test, has_instrumentation_test

# === ANALYZE FILES ===
rows = []
for filename in os.listdir(GRADLE_DIR):
    if filename.endswith('++build.gradle') or filename.endswith('++build.gradle.kts'):
        full_path = os.path.join(GRADLE_DIR, filename)
    try:
        # Split into base and filename (e.g., username.project__CI++build.gradle)
        base, file_part = filename.split("++", 1)
        repo_info, ci_platform = base.split("__", 1)

        full_name = repo_info  # e.g., username.project

        has_unit, has_instrument = detect_tests(full_path)

        print(f"✅ {filename} | Full Name: {full_name} | CI Platform: {ci_platform} | Unit: {has_unit} | Instr: {has_instrument}")

        rows.append({
            'full_name': full_name,
            'ci_platform_build': ci_platform,
            'gradle_file': filename,
            'has_build_unit_test': has_unit,
            'has_build_instrumentation_test': has_instrument
        })
    except Exception as e:
        print(f"❌ Error processing {filename}: {e}")

# === SAVE TO CSV ===
df = pd.DataFrame(rows)
df.to_csv(OUTPUT_CSV, index=False)
print(f"✅ Saved results to: {OUTPUT_CSV}")
```
