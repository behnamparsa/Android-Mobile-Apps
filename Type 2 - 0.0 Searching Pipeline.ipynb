```python
import papermill as pm
import os

# Folder containing your notebooks
notebook_folder = r"C:\GitHub\Android-Mobile-Apps"

# List of notebook files in order
notebooks = [
    "Type 2 - 1.1. Searching Repo_general_qualifiers API Search.ipynb",
    "Type 2 - 1.2. Searching Repo_general_qualifiers REST check.ipynb", 
    "Type 2 - 1.3. Searching Repo_keyword detection.ipynb",
    "Type 2 - 1.4. Searching Repo_keyword verification.ipynb",
    "Type 2 - 1.5. Searching Repo_manifest+activity.ipynb",
    "Type 2 - 1.6. Searching Repo_removal_keyword.ipynb"
]

# Loop through and execute each notebook
for nb in notebooks:
    input_path = os.path.join(notebook_folder, nb)
    output_path = os.path.join(notebook_folder, f"executed_{nb}")
    print(f"🚀 Running {nb}...")

    try:
        pm.execute_notebook(
            input_path,
            output_path,
            kernel_name="python3"
        )
        print(f"✅ {nb} executed successfully.\n")
    except Exception as e:
        print(f"❌ Error while executing {nb}: {e}")
        break  # Stop pipeline on error
```
