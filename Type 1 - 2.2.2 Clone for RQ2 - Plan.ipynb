```python
import json, os

src = "Type 1 - 2.2.1 Clone for RQ2.py"  # <-- the mis-saved file
with open(src, "r", encoding="utf-8") as f:
    head = f.read(4096)
    f.seek(0); txt = f.read()

if head.lstrip().startswith("{") and '"cells"' in head:
    nb = json.loads(txt)
    lines = []
    for c in nb.get("cells", []):
        if c.get("cell_type") == "code":
            # pull raw code lines from each code cell
            lines.extend(c.get("source", []))
            if lines and lines[-1] and not lines[-1].endswith("\n"):
                lines[-1] += "\n"
            lines.append("\n")
    out = src.replace(".py","") + ".fixed.py"
    with open(out, "w", encoding="utf-8") as g:
        g.write("".join(lines))
    print("Wrote:", out)
else:
    print("This file doesn’t look like a notebook JSON.")
```
