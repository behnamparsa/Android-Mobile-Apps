```python
#check the token limit

import os
import requests
from dotenv import load_dotenv

# === Load tokens from .env ===
load_dotenv("All_Tokens.env")

tokens = {
    "GITHUB_TOKEN": os.getenv("GITHUB_TOKEN")

}

def check_rate_limit(token_name, token_value):
    headers = {"Authorization": f"token {token_value}"}
    response = requests.get("https://api.github.com/rate_limit", headers=headers)
    
    if response.status_code == 200:
        rate_info = response.json()["rate"]
        print(f"✅ {token_name}:")
        print(f"   Remaining: {rate_info['remaining']} / {rate_info['limit']}")
        print(f"   Resets at: {rate_info['reset']}\n")
    else:
        print(f"❌ {token_name}: Failed ({response.status_code}) - {response.text}\n")

# === Run for each token ===
for name, value in tokens.items():
    if value:
        check_rate_limit(name, value)
    else:
        print(f"⚠️ {name} is not set in All_Tokens.env\n")
```
