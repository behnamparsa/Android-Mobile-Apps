```python
import os
import pandas as pd

# === CONFIGURATION ===
CONFIG_DIR = r"F:\Android_Mobile_App\Initial Dataset\Config Files"
OUTPUT_CSV = r"F:\Android_Mobile_App\Initial Dataset\0.0-User_Project_List.csv"
os.makedirs(os.path.dirname(OUTPUT_CSV), exist_ok=True)

# === Extract metadata from all files ===
all_data = []

for root, _, files in os.walk(CONFIG_DIR):
    for file_name in files:
        parts = file_name.split('.')
        if len(parts) >= 2:
            username = parts[0]
            projectname = parts[1]
            full_name = f"{username}.{projectname}"
        else:
            username = None
            projectname = None
            full_name = None

        is_yml_or_yaml = file_name.lower().endswith('.yml') or file_name.lower().endswith('.yaml')

        all_data.append({
            'username': username,
            'projectname': projectname,
            'full_name': full_name,
            'file_name': file_name,
            'is_yml_or_yaml': is_yml_or_yaml
        })

# === Save to CSV ===
df = pd.DataFrame(all_data)
df.to_csv(OUTPUT_CSV, index=False)

print(f"✅ Done. Total files processed: {len(df)} — saved to {OUTPUT_CSV}")
```

```python
import os

# === CONFIG ===
CONFIG_DIR = r"F:\Android_Mobile_App\Initial Dataset\Config Files"

# === Count all files ===
total_files = 0
yml_files = 0
all_filenames = []

for root, _, files in os.walk(CONFIG_DIR):
    for file in files:
        total_files += 1
        all_filenames.append(file)
        if '.yml' in file.lower():
            yml_files += 1

print(f"📁 Total files found: {total_files}")
print(f"✅ Files containing '.yml': {yml_files}")

# Print out edge cases
suspicious = [f for f in all_filenames if '.yml' not in f.lower()]
print(f"⚠️ Suspicious files (not containing .yml): {len(suspicious)}")
```
