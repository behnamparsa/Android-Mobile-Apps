```python
import os
import pandas as pd

# === PATHS ===
BASE_DIR   = r"C:\Android Mobile App\Step3_Instr_Testing_Analysis\Type_1\Aug_10\ICST2026"
IN_CSV     = os.path.join(BASE_DIR, "5_Total_Repo.csv")  # your final dataset
OUT_DIR    = os.path.join(BASE_DIR, "Random Sample")
OUT_CSV    = os.path.join(OUT_DIR, "5_Total_Repo_RandomSample_68.csv")

SAMPLE_N    = 68
RANDOM_SEED = 42  # reproducible sampling seed
INDEX_COL   = f"Random_Seed{RANDOM_SEED}"  # column name requested

def main():
    os.makedirs(OUT_DIR, exist_ok=True)

    df = pd.read_csv(IN_CSV, dtype=str).fillna("")

    if "full_name" not in df.columns:
        raise ValueError("Expected a 'full_name' column in the input dataset.")

    # Sample UNIQUE repos by full_name (one row per repo)
    repo_df = df.drop_duplicates(subset=["full_name"], keep="first").reset_index(drop=True)

    # Add reproducible, stable row index column at the BEGINNING
    # (If you later sort by this column, the order is deterministic for this saved file.)
    #repo_df.insert(0, INDEX_COL, range(1, len(repo_df) + 1))  # 1-based index

    if len(repo_df) < SAMPLE_N:
        raise ValueError(f"Only {len(repo_df)} unique repos found, cannot sample {SAMPLE_N}.")

    # Sample using the fixed seed
    sample = repo_df.sample(n=SAMPLE_N, random_state=RANDOM_SEED).reset_index(drop=True)

    sample.to_csv(OUT_CSV, index=False, encoding="utf-8-sig")
    print(f"Saved random sample: {OUT_CSV} (rows={len(sample)} unique repos)")
    print(f"Index column '{INDEX_COL}' added as first column (1..{len(repo_df)}).")

if __name__ == "__main__":
    main()
```
