STEP 1: Search for Android Repositories

```python
import requests
import time
import pandas as pd

# Configuration
GITHUB_TOKEN = 'your_token_here'  # replace with your real token
HEADERS = {'Authorization': f'token {GITHUB_TOKEN}'}
BASE_URL = "https://api.github.com/search/repositories"
QUERY = "android language:java"  # or use kotlin
PER_PAGE = 100
MAX_PAGES = 10  # 10 pages * 100 results = 1000 projects max

repos = []

for page in range(1, MAX_PAGES + 1):
    params = {
        "q": QUERY,
        "sort": "stars",
        "order": "desc",
        "per_page": PER_PAGE,
        "page": page
    }
    print(f"🔍 Fetching page {page}")
    response = requests.get(BASE_URL, headers=HEADERS, params=params)

    if response.status_code == 200:
        items = response.json().get("items", [])
        for repo in items:
            repos.append({
                "full_name": repo["full_name"],
                "html_url": repo["html_url"],
                "clone_url": repo["clone_url"],
                "stars": repo["stargazers_count"],
                "forks": repo["forks_count"],
                "watchers": repo["watchers_count"],
                "open_issues": repo["open_issues_count"],
                "language": repo["language"],
                "size_kb": repo["size"],
                "created_at": repo["created_at"],
                "updated_at": repo["updated_at"],
                "pushed_at": repo["pushed_at"]
            })
    else:
        print(f"⚠️ GitHub API Error: {response.status_code} - {response.text}")
        break

    time.sleep(2)  # prevent hitting rate limits

# Save to CSV for review
df = pd.DataFrame(repos)
df.to_csv("Top_Android_Repos.csv", index=False)
print(f"✅ Saved {len(df)} repositories to Top_Android_Repos.csv")
```
