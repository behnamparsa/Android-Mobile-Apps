```python
import pandas as pd

# === File paths ===
input_path = r"C:\Android Mobile App\Step1_URL_Search\Type_2_Searching_Pipeline\step6_removal_keyword_check_output.csv"
output_path = r"C:\Android Mobile App\Step1_URL_Search\Type_2_Searching_Pipeline\step7_confidence_output.csv"

# === Load dataset ===
df = pd.read_csv(input_path)

# === Add confidence column ===
df["confidence"] = "N/A"

# === Loop over all rows ===
for i, row in df.iterrows():
    if row["Valid_Repo"] != "yes":
        df.at[i, "confidence"] = "none"
        continue

    metadata = row["android_metadata_match"] == "yes"
    readme = row["android_in_readme"] == "yes"
    has_activity = row["has_activity"] == "yes"
    standard_manifest = row["standard_manifest"] == "yes"

    if metadata and (has_activity or standard_manifest):
        df.at[i, "confidence"] = "high"
    elif not metadata and readme and has_activity:
        df.at[i, "confidence"] = "medium"
    elif readme and not has_activity:
        df.at[i, "confidence"] = "low"
    else:
        df.at[i, "confidence"] = "none"

# === Save final output ===
df.to_csv(output_path, index=False)
print(f"✅ Step 7 complete. Confidence scores saved to: {output_path}")
```
