```python
import os
import glob
import pandas as pd

# === Paths ===
CLONED_DIR = "Cloned_Repos"
results = []

for root, dirs, files in os.walk(CLONED_DIR):
    # Check for .yml CI configs
    ymls = glob.glob(os.path.join(root, ".github", "workflows", "*.yml"))
    has_ci = bool(ymls)

    # Check for build.gradle with androidTest
    gradles = glob.glob(os.path.join(root, "**", "build.gradle"), recursive=True)
    has_instrumentation = False
    for gradle in gradles:
        try:
            with open(gradle, "r", encoding="utf-8") as f:
                content = f.read().lower()
                if "androidtest" in content or "espresso" in content:
                    has_instrumentation = True
                    break
        except:
            continue

    if ".git" in dirs:
        repo_name = os.path.relpath(root, CLONED_DIR)
        results.append({
            "repo": repo_name,
            "has_ci_yml": has_ci,
            "has_instrumentation": has_instrumentation
        })

df = pd.DataFrame(results)
df.to_csv("Android_Repos_Verified.csv", index=False)
print("✅ Saved results to Android_Repos_Verified.csv")
print(df.head())
```
