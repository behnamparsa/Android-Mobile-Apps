```python
import os
import pandas as pd

# === CONFIGURATION ===
GRADLE_DIR = r"D:\Android_Mobile_App\AndroidProject_4th\6.2-Shallow_Clone\BuildInfo"
OUTPUT_CSV = r"D:\Android_Mobile_App\AndroidProject_4th\6.2-Shallow_Clone\Analysis Output\7.1-Gradle_List_4th_attempt_ShallowC.csv"

# === TEST DETECTION LOGIC ===
def detect_tests(file_path):
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read().lower()
        has_unit_test = any(keyword in content for keyword in ['junit', 'testimplementation'])
        has_instrumentation_test = any(keyword in content for keyword in ['androidtestimplementation', 'espresso', 'uiautomator'])
        return has_unit_test, has_instrumentation_test

# === ANALYZE FILES ===
rows = []
for filename in os.listdir(GRADLE_DIR):
    if filename.endswith('__build.gradle') or filename.endswith('__build.gradle.kts'):
        full_path = os.path.join(GRADLE_DIR, filename)
        try:
            parts = filename.split("__")

            # ✅ Extract full_name (between first and third '__')
            if len(parts) >= 3:
                full_name = f"{parts[1]}__{parts[2]}"
            else:
                full_name = filename  # fallback

            # ✅ Extract CI platform (between third and fourth '__')
            if len(parts) >= 4:
                ci_platform = parts[3]
            else:
                ci_platform = "Unknown"

            has_unit, has_instrument = detect_tests(full_path)

            print(f"✅ {filename} | Full Name: {full_name} | CI Platform: {ci_platform} | Unit: {has_unit} | Instr: {has_instrument}")

            rows.append({
                'full_name': full_name,
                'ci_platform_build': ci_platform,
                'gradle_file': filename,
                'has_local_unit_test': has_unit,
                'has_local_instrumentation_test': has_instrument
            })
        except Exception as e:
            print(f"❌ Error processing {filename}: {e}")

# === SAVE TO CSV ===
df = pd.DataFrame(rows)
df.to_csv(OUTPUT_CSV, index=False)
print(f"✅ Saved results to: {OUTPUT_CSV}")
```
