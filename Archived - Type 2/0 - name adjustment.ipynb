```python
from pathlib import Path
import json

src_dir = Path(r"C:\Android Mobile App\Step3_Instr_Testing_Analysis\Type_1\Aug_10\ICST2026\2 - RQ2\Mine_Full")
dst_dir = Path(r"C:\Android Mobile App\Step3_Instr_Testing_Analysis\Type_1\Aug_10\ICST2026\2 - RQ2\Mine_Full_2")

dst_dir.mkdir(parents=True, exist_ok=True)

for src_file in src_dir.rglob("*.json"):  # recursive; use src_dir.glob("*.json") for non-recursive
    rel = src_file.relative_to(src_dir)

    # Preserve subfolders, but change filename: name.json -> name.default_refs.json
    dst_path_base = dst_dir / rel
    dst_file = dst_path_base.with_name(f"{dst_path_base.stem}.default_refs{dst_path_base.suffix}")
    dst_file.parent.mkdir(parents=True, exist_ok=True)

    # Read + write JSON (re-serializes the JSON)
    with src_file.open("r", encoding="utf-8") as f:
        data = json.load(f)

    with dst_file.open("w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

print("Done.")
```
