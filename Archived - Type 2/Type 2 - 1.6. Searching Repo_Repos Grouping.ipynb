```python
import pandas as pd

# === File paths ===
input_path = r"C:\Android Mobile App\Step1_URL_Search\Type_2_Searching_Pipeline_July 16\step5_removal_keyword_check_output.csv"
output_path = r"C:\Android Mobile App\Step1_URL_Search\Type_2_Searching_Pipeline_July 16\step6_confidence_output.csv"

# === Load dataset ===
#df = pd.read_csv(input_path)
df = pd.read_csv(input_path, dtype={'rest_reason': str})


# === Add confidence column ===
df["confidence"] = "N/A"

# === Loop and assign confidence levels ===
for i, row in df.iterrows():
    # Step 1: Must be valid in Step 5 and have manifest + activity
    if (
        row.get("Valid_Repo_Step5", "no") != "yes"
        or row.get("has_manifest", "no") != "yes"
        or row.get("has_activity", "no") != "yes"
    ):
        df.at[i, "confidence"] = "none"
        continue

    # Step 2: Determine confidence based on metadata/readme match
    metadata = row.get("android_metadata_match", "") == "yes"
    readme = row.get("android_in_readme", "") == "yes"

    if metadata:
        df.at[i, "confidence"] = "high"
    elif readme:
        df.at[i, "confidence"] = "medium"
    else:
        df.at[i, "confidence"] = "low"

# === Save final output ===
df.to_csv(output_path, index=False)
print(f"✅ Step 6 complete. Confidence scores saved to: {output_path}")
```
