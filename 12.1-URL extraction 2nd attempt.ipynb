```python
import os
import pandas as pd

# Input and output directories
config_dir = r"F:\Android_Mobile_App\Step by Step Search\Config Files"
output_dir = r"F:\Android_Mobile_App\2nd_attempt"
os.makedirs(output_dir, exist_ok=True)

# Extract GitHub URLs based on the filename format
repo_urls = []
for file in os.listdir(config_dir):
    if file.endswith(".yml"):
        name_parts = file.split(".")
        if len(name_parts) >= 2:
            username = name_parts[0]
            projectname = name_parts[1]
            repo_url = f"https://github.com/{username}/{projectname}"
            repo_urls.append((username, projectname, repo_url))

# Remove duplicates based on repo_url
df_urls = pd.DataFrame(repo_urls, columns=["username", "project", "repo_url"])
df_urls = df_urls.drop_duplicates(subset="repo_url")

# Save to CSV
output_csv_path = os.path.join(output_dir, "Unique_Repo_URLs.csv")
df_urls.to_csv(output_csv_path, index=False)

print(f"✅ Saved {len(df_urls)} unique repository URLs to:\n{output_csv_path}")
```
