```python
import pandas as pd
import os
import subprocess
from dotenv import load_dotenv

# === CONFIG ===
BASE_DIR = r"F:\Android_Mobile_App\AndroidProject_4th"
INPUT_CSV = os.path.join(BASE_DIR, "Android_Repos_MultiRange.csv")
OUTPUT_DIR = os.path.join(BASE_DIR, "Cloned_Repos")

# === Load GitHub token and START_NUMBER ===
load_dotenv("All_Tokens.env")
GITHUB_TOKEN = os.getenv("GITHUB_TOKEN")
START_NUMBER = int(os.getenv("START_NUMBER", "0"))

# === Create output folder ===
os.makedirs(OUTPUT_DIR, exist_ok=True)

# === Load and sort CSV ===
df = pd.read_csv(INPUT_CSV)
df = df.sort_values(by="stars", ascending=False).reset_index(drop=True)

# === Shallow clone with optional token ===
current_index = START_NUMBER

while current_index < len(df):
    row = df.iloc[current_index]
    full_name = row["full_name"]
    clone_url = row["clone_url"]

    # If you have a token, insert it in the URL
    if GITHUB_TOKEN:
        if clone_url.startswith("https://"):
            parts = clone_url.split("https://")
            clone_url = f"https://{GITHUB_TOKEN}@{parts[1]}"

    dest = os.path.join(OUTPUT_DIR, full_name.replace("/", "."))
    if os.path.exists(dest):
        print(f"✅ Already exists: {dest} [index {current_index}]")
    else:
        print(f"🚀 [{current_index}] Cloning {full_name} ...")
        subprocess.run(["git", "clone", "--depth", "1", clone_url, dest])

    current_index += 1

    # Update START_NUMBER in the .env file on each loop
    with open("All_Tokens.env", "r") as f:
        lines = f.readlines()

    with open("All_Tokens.env", "w") as f:
        for line in lines:
            if line.startswith("START_NUMBER="):
                f.write(f"START_NUMBER={current_index}\n")
            else:
                f.write(line)

print(f"\n✅ Finished cloning up to index {current_index - 1}. Updated START_NUMBER saved.")
```
