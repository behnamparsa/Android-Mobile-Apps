```python
import os
import subprocess
from dotenv import load_dotenv

# === CONFIG ===
ENV_FILE = 'All_tokens.env'  # Use your .env file with GITHUB_TOKEN
REPO_URL = "https://github.com/myofficework000/Jetpack-Compose-All-in-one-Guide.git"
CLONE_DIR = r"C:\Android Mobile App\Single Repo"  # adjust to your desired path

# === LOAD TOKEN ===
load_dotenv(ENV_FILE)
GITHUB_TOKEN = os.getenv('GITHUB_TOKEN')

if not GITHUB_TOKEN:
    raise ValueError("❌ GitHub token not found in .env")

# === Inject token into URL ===
url_parts = REPO_URL.split("https://")
token_url = f"https://{GITHUB_TOKEN}@{url_parts[1]}"

# === Ensure target folder exists ===
os.makedirs(CLONE_DIR, exist_ok=True)

# === Clone the repository ===
try:
    subprocess.run(
        ["git", "clone", "--depth", "1", "--single-branch", token_url],
        cwd=CLONE_DIR,
        check=True
    )
    print(f"✅ Clone complete: {REPO_URL}")
except subprocess.CalledProcessError as e:
    print(f"❌ Clone failed: {e}")
```
