```python
import zipfile
import os

zip_path = 'F:\Android_Mobile_App\Step by Step Search/Config Files.zip'
extract_dir = 'F:\Android_Mobile_App\Step by Step Search/config_extracted'

# Unzip
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# List contents
for root, dirs, files in os.walk(extract_dir):
    for f in files:
        print(os.path.join(root, f))
```
