```python
import os
from pathlib import Path
import shutil

# === CONFIG ===
source_folder = Path(r"F:\Android_Mobile_App\AndroidProjects_FullorShallow_Shallow_default_Complete\BuildInfo")
target_folder = source_folder.parent / "Fixed Names"
target_folder.mkdir(exist_ok=True)

# === RENAME AND COPY FILES ===
for file in source_folder.glob("*.*"):
    if file.is_file():
        parts = file.name.split(".")

        # Make sure there's at least: username.project.filename.ext
        if len(parts) >= 4:
            username = parts[0]
            project = parts[1]
            filename = ".".join(parts[-2:])  # last two parts (e.g., name.ext)
            new_name = f"{username}.{project}.{filename}"
            shutil.copy2(file, target_folder / new_name)
            print(f"✅ Renamed: {file.name} → {new_name}")
        else:
            print(f"⚠️ Skipped: {file.name} (unexpected format)")
```
