```python
import os
import pandas as pd

root_dir = 'F:\Android_Mobile_App\Step by Step Search\Config Files'
project_data = []

for project_name in os.listdir(root_dir):
    project_path = os.path.join(root_dir, project_name)
    if not os.path.isdir(project_path):
        continue

    latest_yml = None
    latest_time = 0

    # Search for .yml/.yaml files recursively
    for dirpath, _, filenames in os.walk(project_path):
        for filename in filenames:
            if filename.endswith(('.yml', '.yaml')):
                file_path = os.path.join(dirpath, filename)
                mtime = os.path.getmtime(file_path)
                if mtime > latest_time:
                    latest_time = mtime
                    latest_yml = file_path

    if latest_yml:
        project_data.append({
            'project_name': project_name,
            'latest_yml_path': latest_yml,
            'last_modified': pd.to_datetime(latest_time, unit='s')
        })

# Convert to DataFrame
df_yml = pd.DataFrame(project_data)
import ace_tools as tools; tools.display_dataframe_to_user(name="Latest CI Config Files", dataframe=df_yml)
```
